# Interactive 3D MNIST embedding

Explore one 3D embedding from `IteratedMSTEmbedding` on the same standardized MNIST digits as the parameter sweep notebook. The embedding uses `n_msts=8` and Plotly for rotation, zoom, and hover labels. Depth fog fades points farther from the current camera and updates as you rotate. The stratified sample defaults to 2,000 rows; change `SAMPLE_SIZE` or set it to `None` for all 70,000 rows.

The estimator defaults to two dimensions; this notebook sets `n_components=3` for the 3D plot. Install plotting dependencies with `pip install -e ".[notebook]"`. The first data cell downloads MNIST from OpenML if it is not cached.


```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import sys

repo_root = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / "notebooks" / "digits_sweep.py").is_file()),
    None,
)
if repo_root is not None:
    sys.path.insert(0, str(repo_root / "notebooks"))

from digits_sweep import load_mnist_data
from digits_3d import fit_and_plot_mst_3d, render_with_camera_fog
```

## Load a stratified MNIST sample

Features are standardized to zero mean and unit variance, matching the parameter sweep notebook.


```python
SAMPLE_SIZE = 4000  # Set to None to use all 70,000 rows.
DATA_RANDOM_STATE = 42
X, labels = load_mnist_data(
    sample_size=SAMPLE_SIZE,
    random_state=DATA_RANDOM_STATE,
)
print(f"Samples: {X.shape[0]}, features: {X.shape[1]}")
```

## Fit and visualize the 3D embedding

The score table reports trustworthiness, five-fold 5-NN accuracy, elapsed time, and the selected compute device. Drag the plot to rotate it; camera-aware fog updates as the view changes. Scroll to zoom, and hover over a point to see its digit label.


```python
N_MSTS = 30
N_EPOCHS = 500

embedding_3d, scores_3d, figure_3d = fit_and_plot_mst_3d(
    X, labels,
    n_msts=N_MSTS,
    n_epochs=N_EPOCHS,
    batch_size=4096,
    learning_rate=0.05,
    negative_ratio=4,
    lambda_rep=1.0,
    epsilon=1e-4,
    random_state=DATA_RANDOM_STATE,
    device="auto",
)
display(render_with_camera_fog(figure_3d, embedding_3d, labels))
```
