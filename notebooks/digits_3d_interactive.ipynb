# Interactive 3D MNIST embedding

Explore one 3D embedding from `IteratedMSTEmbedding` on the same standardized MNIST digits as the parameter sweep notebook. The embedding uses `n_msts=8`. Set `RANK_WEIGHT_EXPONENT` to control how quickly later MST ranks lose influence: 0 gives equal weights, and 1 gives inverse-rank weights. The chart below is a standard interactive Plotly 3D scatter: rotate, zoom, and hover to inspect the embedding. The stratified sample defaults to 4,000 rows; change `SAMPLE_SIZE` or set it to `None` for all 70,000 rows.

The estimator defaults to two dimensions; this notebook sets `n_components=3` for the 3D plot. Install plotting dependencies with `pip install -e ".[notebook]"`. The first data cell downloads MNIST from OpenML if it is not cached.


```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import sys

repo_root = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / "notebooks" / "digits_sweep.py").is_file()),
    None,
)
if repo_root is not None:
    sys.path.insert(0, str(repo_root / "notebooks"))

from digits_sweep import load_mnist_data
from digits_3d import fit_and_plot_mst_3d
```

## Load a stratified MNIST sample

Features are standardized to zero mean and unit variance, matching the parameter sweep notebook.


```python
SAMPLE_SIZE = 4000  # Set to None to use all 70,000 rows.
DATA_RANDOM_STATE = 42
X, labels = load_mnist_data(
    sample_size=SAMPLE_SIZE,
    random_state=DATA_RANDOM_STATE,
)
print(f"Samples: {X.shape[0]}, features: {X.shape[1]}")
```

## Fit and visualize the 3D embedding

The score table reports trustworthiness, five-fold 5-NN accuracy, elapsed time, and the selected compute device. The Plotly chart can be rotated, zoomed, and hovered in the notebook.


```python
embedding_3d, scores_3d, figure_3d = fit_and_plot_mst_3d(
    X, labels,
    n_msts=12,
    rank_weight_exponent=0.75,
    n_epochs=500,
    batch_size=4096,
    learning_rate=0.05,
    negative_ratio=4,
    lambda_rep=1.0,
    epsilon=1e-4,
    random_state=DATA_RANDOM_STATE,
    device="auto",
)

display(figure_3d)
```
