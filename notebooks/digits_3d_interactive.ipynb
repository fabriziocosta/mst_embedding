# Interactive 3D MNIST embeddings

Explore the same standardized MNIST digits with interactive 3D t-SNE. Each Plotly panel has its own camera: drag to rotate, scroll to zoom, and hover to see the digit label. The default stratified sample is 2,000 rows; change `SAMPLE_SIZE` below to use a different number or set it to `None` for all 70,000 rows.

Install the notebook extras first with `pip install -e '.[notebook]'`. The first data cell downloads MNIST from OpenML if it is not already cached.


```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import sys

repo_root = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / "notebooks" / "digits_sweep.py").is_file()),
    None,
)
if repo_root is not None:
    sys.path.insert(0, str(repo_root / "notebooks"))

from digits_sweep import load_mnist_data
from digits_3d import run_tsne_3d_comparisons
```

## Load a stratified MNIST sample

Features are standardized to zero mean and unit variance, matching the parameter sweep notebook.


```python
SAMPLE_SIZE = 2000  # Set to None to use all 70,000 rows.
DATA_RANDOM_STATE = 42
X, labels = load_mnist_data(
    sample_size=SAMPLE_SIZE,
    random_state=DATA_RANDOM_STATE,
)
print(f"Samples: {X.shape[0]}, features: {X.shape[1]}")
```

## Compare three t-SNE perplexities

The settings match the 2D comparison: perplexities 5, 30, and 50, with the same seed and PCA initialization. Each embedding is computed in 3D and shown in an interactive panel.


```python
TSNE_CONFIGS = [
    {"name": "Local", "perplexity": 5},
    {"name": "Default", "perplexity": 30},
    {"name": "Broader", "perplexity": 50},
]

tsne_3d_results, figure = run_tsne_3d_comparisons(
    X,
    labels,
    TSNE_CONFIGS,
    n_iterations=1000,
    random_state=DATA_RANDOM_STATE,
)
figure
```
